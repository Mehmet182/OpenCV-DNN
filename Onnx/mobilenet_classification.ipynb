```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
net = cv2.dnn.readNetFromONNX("mobilenetv2-10.onnx")  # kendi yolunu yaz
```

```python
with open("classes.txt", "r") as f:
    labels = [line.strip() for line in f.readlines()]
```

```python
image = cv2.imread("cat.jpg")  # kendi görselini yaz
h, w = image.shape[:2]
```

```python
blob = cv2.dnn.blobFromImage(image, scalefactor=1/255.0, size=(224,224), swapRB=True, crop=False)
net.setInput(blob)
```

```python
out = net.forward()   # shape (1,1000)
scores = out[0]
```

```python
cls_id = int(np.argmax(scores))
conf = float(scores[cls_id])
```

```python
label = labels[cls_id] if cls_id < len(labels) else str(cls_id)
```

```python
text = f"{label}: {conf:.2f}"
cv2.putText(image, text, (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (0,255,0), 2)

# 8. Sonucu göster
cv2.imshow("Result", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
