```python
import cv2
import numpy as np
import os
```

```python
prototxt_path = "ResNet-50-deploy.prototxt"
caffemodel_path = "ResNet-50-model.caffemodel" # weight
labels_path="synset_words.txt"
image_path="cat.jpg"
```

```python
print("Current Working Directory:", os.getcwd())
print("config Dosyası Var mı?:", os.path.exists(prototxt_path))
print("class Dosyası Var mı?:", os.path.exists(caffemodel_path))
print("label Dosyası Var mı?:", os.path.exists(labels_path))
print("image Dosyası Var mı?:", os.path.exists(image_path))
```

```python
# --- Class isimlerini oku ---
with open(labels_path, "r") as f:
    labels = [line.strip() for line in f.readlines()]
```

```python
net = cv2.dnn.readNetFromCaffe(prototxt_path, caffemodel_path)
```

```python
image = cv2.imread(image_path)
(h, w) = image.shape[:2]
```

```python
blob = cv2.dnn.blobFromImage(image, 
                             scalefactor=1.0, 
                             size=(224, 224), 
                             mean=(104, 117, 123))
```

```python
net.setInput(blob)
preds = net.forward()
```

```python
# --- En büyük skor ---
idx = np.argmax(preds[0])
confidence = preds[0][idx]
label = f"{labels[idx]}, {confidence:.2f}"
```

```python
# --- Resmin üzerine yaz ---
cv2.putText(image, label, (10, 30),
            cv2.FONT_HERSHEY_SIMPLEX, 0.8,
            (0, 255, 0), 2)

# --- Görüntüyü göster ---
cv2.imshow("Classification", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
