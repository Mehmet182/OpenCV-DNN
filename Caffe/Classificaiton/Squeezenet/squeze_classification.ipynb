```python
import cv2
import numpy as np
import os
```

```python


prototxt_path = "Squeze.prototxt"
caffemodel_path = "squeezenet_v1.0.caffemodel"
labels_path = "synset_words.txt"
image_path="1261.jpg"

# --- Label dosyasını oku ---
with open(labels_path, "r") as f:
    labels = [line.strip() for line in f.readlines()]

# --- Modeli yükle ---
net = cv2.dnn.readNetFromCaffe(prototxt_path, caffemodel_path)

# --- Görsel yükle ---
image = cv2.imread(image_path)
(h, w) = image.shape[:2]

# --- Blob oluştur ---
# SqueezeNet de ImageNet mean değerleri (104,117,123) ile eğitildi
blob = cv2.dnn.blobFromImage(image,
                             scalefactor=1.0,
                             size=(227, 227),   # SqueezeNet için genelde 227x227
                             mean=(104, 117, 123), swapRB=False,          # Caffe modellerinde False
                             crop=False)

# --- Forward pass ---
net.setInput(blob)
preds = net.forward()

# --- En büyük skor ---
idx = np.argmax(preds[0])
confidence = float(preds[0][idx])   # <-- düzeltme
label = f"{labels[idx]}, {confidence:.2f}"

# --- Resmin üzerine yaz ---
cv2.putText(image, label, (w//10, h//10),
            cv2.FONT_HERSHEY_SIMPLEX, 0.8,
            (0, 255, 0), 2)


# --- Görüntüyü göster ---
cv2.imshow("SqueezeNet Classification", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
print("Index:", idx)
print("Confidence:", confidence)
print("Label file line idx:", labels[idx])
print("Label file line idx+1:", labels[idx+1])
```
